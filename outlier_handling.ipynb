**Step1: Import necessary libraries**

```python
import numpy as np
import seaborn as sns
```

**Step2: Sorting data in ascending order**

```python
data = [6,2,3,4,5,1,50]
sort_data = np.sort(data)
sort_data
```

**Step3: Calculating Q1, Q2, Q3 and IQR**

```python
Q1 = np.percentile(data, 25, interpolation= 'midpoint')
Q2 = np.percentile(data, 50, interpolation= 'midpoint')
Q3 = np.percentile(data, 75, interpolation= 'midpoint')

print('Q1 25% of given data is ',Q1)
print('Q2 50% of given data is ',Q2)
print('Q3 75% of given data is ',Q3)

IQR = Q3 - Q1
print('Interquartile range is ',IQR)
```

**Step4: Find the lower and upper limits**

```python
low_lim = Q1 - 1.5*IQR
up_lim = Q3 + 1.5*IQR

print('Lower limit is ',low_lim)
print('Upper limit is ',up_lim)
```

**Step5: Identify the outliers**

```python
outlier =[]
for x in data:
    if ((x> up_lim) or (x<low_lim)):
         outlier.append(x)
print(' outlier in the dataset is', outlier)
```

**Step6: Plot the box plot to highlight outliers**

```python
sns.boxplot(data)
```
